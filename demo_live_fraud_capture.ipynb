```python
import subprocess
import os

service_script_path = os.path.join(os.getcwd(), 'scripts/live/realtime_inference_service.py')

print(f"Starting real-time inference service in the background: {service_script_path}")
print("Monitor terminal output (where Jupyter kernel is running) for live predictions and check 'output/flagged_transactions.csv' for logged fraud.")

# Start the script in the background.
# Removing stdout/stderr redirection to see potential errors directly in notebook output.
process = subprocess.Popen(['python3', service_script_path]) # Removed stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL

print(f"Service started with PID: {process.pid}")
print("**Important:** Output will now appear directly in this cell's output or the terminal running Jupyter. To terminate, run the next cell.")
```

```python
# Read the log file here to see its current contents:
import pandas as pd
import os

flagged_transactions_file = os.path.join(os.getcwd(), 'output/flagged_transactions.csv')

try:
    df_flagged = pd.read_csv(flagged_transactions_file)
    print("\n--- Current Flagged Transactions Log (output/flagged_transactions.csv) ---")
    print(df_flagged.tail())
except FileNotFoundError:
    print(f"Log file not found: {flagged_transactions_file}. No fraud detected yet or service not started.")
except pd.errors.EmptyDataError:
    print(f"Log file is empty: {flagged_transactions_file}. No fraud detected yet.")

```

```python
import subprocess

# Find the PID of the running script and kill it
pids = subprocess.run(
    ['pgrep', '-f', 'python3 .*scripts/live/realtime_inference_service.py'], 
    capture_output=True, text=True
).stdout.strip()

if pids:
    subprocess.run(['kill', pids.split('\n')[0]]) # Only kill the first matching process if multiple are found
    print(f"Terminated real-time inference service with PID(s): {pids}")
else:
    print("No running real-time inference service found.")

# Also check if there are any zombie processes or hanging children
subprocess.run(['pkill', '-f', 'python3 .*scripts/live/realtime_inference_service.py'], check=False)

print("Service termination command sent.")
```
